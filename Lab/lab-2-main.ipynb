```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

model_id = "mistralai/Mistral-Nemo-Instruct-2407"
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id, torch_dtype=torch.bfloat16, device_map="auto")
```

```python
def model_answer(question, max_length = 100, num_return_sequences=1):
    inputs = tokenizer(
                question,
                return_tensors="pt",
    )
    outputs = model.generate(
        **inputs,
        num_return_sequences=num_return_sequences,
        max_length = max_length,
        max_new_tokens=1000,
        top_k=50,
        top_p=0.95,
        temperature=0.7)
    
    return [tokenizer.decode(output, skip_special_tokens=True) for output in outputs]
```

```python
question1 = "Where is Tips Hindawi University located?"
answer1 = model_answer(question1)
```

```python
question2 = "Does the university offer online programs?"
answer2 = model_answer(question2)
```

```python
question3 = "Is there financial aid for international students?"
answer3 = model_answer(question3)
```

```python
question4 = "What languages are used for instruction?"
answer4 = model_answer(question4)
```

```python
print(answer1[0])
print(answer2[0])
print(answer3[0])
print(answer4[0])
```

```python
from sentence_transformers import SentenceTransformer
from PyPDF2 import PdfReader
import faiss
```

```python
def pdf_to_text(pdf_path):
    read_pdf = PdfReader(pdf_path)
    content = ""
    pages = read_pdf.pages
    for i in range(0,len(pages),1):
        page = read_pdf.pages[i]
        page_content = page.extract_text()
        content += page_content + "\n"
    return content
```

```python
def chunk_text(pdf_content, chunk_size=250, overlap=50):
    words = pdf_content.split()
    chunks = []
    for i in range(0, len(words), chunk_size - overlap):
        chunk = " ".join(words[i:i+chunk_size])
        chunks.append(chunk)
    return chunks
        
```

```python
def chunks_embedding(chunks):
    model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
    embeddings = model.encode(chunks, convert_to_numpy = True)
    return model, embeddings
```

```python
def create_faiss_index(embeddings):
    d = embeddings.shape[1]
    index = faiss.IndexFlatL2(d)
    index.add(embeddings)
    return index
```

```python
def search_index(question, model, index, chunks, k=3):
    embedding_question = model.encode([question], convert_to_numpy = True)
    D, I = index.search(embedding_question, k)
    return [chunks[i] for i in I[0]]
```

```python
pdf_path = "/kaggle/input/datasets/void6639/tips-hindawi-uni/Tips Hindawi University Info.pdf"
text_content = pdf_to_text(pdf_path)
chunks = chunk_text(text_content)
embedding_model, embeddings = chunks_embedding(chunks)
index = create_faiss_index(embeddings)
```

```python
question1 = "Where is Tips Hindawi University located?"
question2 = "Does the university offer online programs?"
question3 = "Is there financial aid for international students?"
question4 = "What languages are used for instruction?"
```

```python
q_top = search_index(question1, embedding_model, index, chunks)
prompt = f"Answer the next question: {question1} using the following text:{q_top[0]}"
answer = model_answer(prompt)
```

```python
print(answer[0])
```

```python
questions = [question1, question2, question3, question4]
answers = []
for question in questions:
    q_top = search_index(question, embedding_model, index, chunks)
    prompt = f"Answer the next question: {question} using the following text:{q_top[0]}"
    answer = model_answer(prompt)
    answers.append(answer)
```

```python
for answer in answers:
    print(answer[0] + "\n\n\n\n")
```
